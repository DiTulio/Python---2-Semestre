```python
class Pilha:
    def __init__(self):
        self.itens = [] #Cria a pilha vazia
    def push(self, item):
        self.itens.append(item) #Adiciona no topo
    def pop(self):
        return self.itens.pop() #Remove do topo

#Uso
p = Pilha() #Cria um objetvo de classe Pilha
p.push(10)
```

```python
    #Exemplo2

class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    #Método a ser criado

    def apresentar(self):
        return f"Meu nome é {self.name} e tenho {self.idade} anos!"

#Criar objeto

p1 = Pessoa("Ana", 35)
print(p1.apresentar())

print(p1.__dict__)
```

```python
#Exercicio 3 - Crie uma classe para mostrar uma estrutura com as seguintes variáveis: nome, carro, ano e modelo. Crie um método para mostrar um texto com as quatro variáveis.

class Cliente:
    def __init__(self, nome, carro, ano, modelo):
        self.nome = nome
        self.carro = carro
        self.ano = ano
        self.modelo = modelo

    def informacoes(self):
        return f"O cliente {self.name} possui um {self.carro} {self.ano}, modelo {self.modelo}"

c1 = Cliente("Bruno", "Honda Fit", 2022, "Hatchback")
print(c1.informacoes)

print(c1.__dict__)
```

```python
#Exercício 4 - Calcular a área e perímetro de retângulo. Crie as duas variáveis: base e altura. Crie dois métodos: área e perímetro.

class Calcular:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return (self.base) * (self.altura) / 2

    def perimetro(self):
        return (self.base * 2)  + (self.altura * 2)

c1 = Calcular(40, 20)
print(c1.area)
print(c1.perimetro)

print(c1.__dict__)
```
