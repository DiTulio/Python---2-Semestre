```python
import csv

caminho = "dataset_falhas_maquinas_Sheet1_.csv"

dados = []
with open(caminho, encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo, delimiter=";")
    for linha in leitor:
        dados.append(linha)

# --- Versão 1: usando WHILE ---
resultado_while = []
i = 0
while i < len(dados):
    linha = dados[i]
    ar = float(linha["Temperatura Ar [K]"].replace(",", "."))
    processo = float(linha["Temperatura Processo [K]"].replace(",", "."))
    temp_media = (ar + processo) / 2

    resultado_while.append({
        "UDI": linha["UDI"],
        "Temperatura Ar [K]": ar,
        "Temperatura Processo [K]": processo,
        "Temp-Media": temp_media
    })
    i += 1

print("Resultado com while:")
for linha in resultado_while[:5]:
    print(linha)

# --- Versão 2: usando FOR + IF ---
resultado_for = []
for linha in dados:
    if "Temperatura Ar [K]" in linha and "Temperatura Processo [K]" in linha:
        ar = float(linha["Temperatura Ar [K]"].replace(",", "."))
        processo = float(linha["Temperatura Processo [K]"].replace(",", "."))

        resultado_for.append({
            "UDI": linha["UDI"],
            "Temperatura Ar [K]": ar,
            "Temperatura Processo [K]": processo,
            "Temp-Media": (ar + processo) / 2
        })

print("\nResultado com for + if:")
for linha in resultado_for[:5]:
    print(linha)
```
