```python
import pandas as pd

class Retangulo:

    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    
    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2* (self.base + self.altura)



r1 = Retangulo(10, 30)
r2 = Retangulo(20, 15)
r3 = Retangulo(25, 15)
r4 = Retangulo(8, 4)
r5 = Retangulo(100, 48)

lista_retangulo = [r1, r2, r3, r4, r5]

dados = []

for i in lista_retangulo:
    linha = {
        "Base": i.base,
        "Altura": i.altura,
        "Area": i.area(),
        "Perímetro": i.perimetro(),
    }

    dados.append(linha)

df = pd.DataFrame(dados)
print(df)
```
