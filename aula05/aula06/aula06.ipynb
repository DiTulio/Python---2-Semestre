```python
import pandas as pd
 
class Carro:
    def __init__(self, marca, modelo, ano, cor, tipo):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.cor = cor
        self.tipo = tipo
 
#metodo para exibir informações do carro
    def exibir_informacoes(self):
        return f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}, Cor: {self.cor}, Tipo: {self.tipo}"
 
#criar objetos da classe Carro
carro1 = Carro("Toyota", "Corolla", 2020, "Prata", "Sedan")
carro2 = Carro("Honda", "Civic", 2019, "Preto", "Hatch")
carro3 = Carro("Ford", "Mustang", 2021, "Vermelho", "Esportivo")
 
lista_carros = [carro1, carro2, carro3]
 
dados = [] # 1. lista vazia
 
for carro in lista_carros: # 2. um objeto por vez
    linha = { #monta o dicionario
        "Marca": carro.marca,
        "Modelo": carro.modelo,
        "Ano": carro.ano,
        "Cor": carro.cor,
        "Tipo": carro.tipo
    }
 
    dados.append(linha)
 
dataFrame = pd.DataFrame(dados)
```
