```python
alunos = {
    "Ana": {"n1": 8, "n2": 7},
    "Bruno": {"n1": 5, "n2": 6}
}

for nome, notas in alunos.items():
    for prova, valor in notas.items():
        print(nome, prova, valor)



alunos = {
    "Ana": {"n1": 8, "n2": 7},
    "Bruno": {"n1": 5, "n2": 6}
}

for i, (nome, notas) in enumerate(alunos.items()):
    print(i, nome, notas)
```

```python
carrinho = {}
produto = ""

while produto != "sair":
    produto = input("Produto: ")
    if produto == "sair":
        break
    preco = float(input("Preço: "))
    carrinho[produto] = preco

print(carrinho)
```

```python
#Exercício 1

aluno = {}
nome = ""
notas = {"n1", "n2", "n3"}

while nome != "sair":
    if nome == "sair":
        break
    else:
        nome = input("Digite seu nome: ")
        n1 = float(input("Digite sua primeira nota: "))
        n2 = float(input("Digite sua segunda nota: "))
        n3 = float(input("Digite sua terceira nota: "))
        media = (n1 + n2 + n3) / 3
        if media >= 6:
            print("Aprovado")
        else:
            print("Reprovado")
            aluno[nome] = notas

print(aluno, nome, notas)
```
