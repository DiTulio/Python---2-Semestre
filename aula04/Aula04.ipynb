```python
class Retangulo:

    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    
    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2* (self.base + self.altura)

    def apresentar (self):
        return f"Area - {self.area()}, Perímetro - {self.perimetro()}"


r1 = Retangulo(4, 5)

print(r1.apresentar())
```

```python
#Exercício Crie uma estrutura com função class para calcular o IMC de uma pessoa. Utilize as variáveis nome, peso e altura.

class imc:
    def __init__(self, nome, peso, altura):
        self.nome = nome
        self.peso = peso
        self.altura = altura

    def calcular(self):
        return self.peso / (self.altura * self.altura)

    def apresentar(self):
        return f"Meu nome é {self.nome} e meu imc é {self.peso / (self.altura * self.altura)}"

i1 = imc("Bruno", 25, 160)
print(i1.apresentar())
```
